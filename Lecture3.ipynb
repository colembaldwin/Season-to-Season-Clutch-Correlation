```python
ampl = ampl_notebook(
    modules=["gurobi"],
    license_uuid="90d58544-d14b-46a7-9ad3-c235239af899"
)
```

```python
ampl.reset()
ampl.eval(r"""
var s>=0;
var r>=0;
var f>=0;
var p>=0;
var w>=0;
minimize Cost:
  4*s+5*r+3*f+7*p+6*w;
subject to Constraint1:
  10*s+20*r+10*f+30*p+20*w >=16;
subject to Constraint2:
  5*s+7*r+4*f+9*p+2*w >=10;
subject to Constraint3:
  1*s+4*r+10*f+2*p+1*w >=15;
subject to Constraint4:
  500*s+450*r+160*f+300*p+500*w >=600;
""")
```

```python
ampl.option["solver"] = "gurobi"
ampl.solve()

ampl.eval("display s, r, f, p, w, Cost;")
```
